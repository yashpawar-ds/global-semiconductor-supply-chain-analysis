## Dataset Overview: Summary of Individual Files


1) Semiconductor_trade_flows.csv - This file contains year-wise semiconductor export and import data. It provides detailed information on exporter and importer countries, hardware types, hardware categories, semiconductor supplier companies, and associated geopolitical risk scores.

2) Trade_sanctions_export_controls.csv - This file tracks annual trade sanctions and export control policies between nations. It includes information on sanctioning and targeted countries, the specific types of restrictions imposed, and the estimated financial impact of these restrictions in USD.

3) National_ai_infrastructure_budgets.csv - This file outlines national AI investments across multiple countries. It details budget allocations across public and private sectors, funding for AI data centers, and workforce development initiatives, along with an AI Sovereignty Index measuring each country's self-sufficiency in the AI sector.

4) Energy_grid_capacity.csv - This file details total national power grid capacities and tracks the proportion of electricity consumed by AI data centers. It also provides a breakdown of energy generation sources, including renewables, coal, natural gas, and nuclear power.

5) Macroeconomic_indicators.csv - This file highlights the macroeconomic drivers shaping the global semiconductor supply chain. Key variables include annual national GDP, GDP growth rates, inflation rates, de-dollarization indices, and unemployment rates.

6) company_profiles.csv - This file profiles key semiconductor supplier companies globally. It contains foundational years, core strategic focus areas, annual revenues, and geopolitical exposure metrics for each organization.


*This dataset combines real and synthetic data. Columns flagged is_synthetic indicate rows generated to follow realistic statistical patterns where real data was not publicly available (e.g., trade sanction economic impact estimates). This analysis treats both as equally valid for pattern exploration, but findings should be understood as illustrative of realistic dynamics rather than verified real-world figures.

 *This dataset consists of multiple relational files, each containing critical domain-specific information about the global semiconductor supply chain. Together, these files provide a comprehensive view of the entire ecosystem

The correlation between sanction count and trade volume is 0.16, a very 
weak relationship. China's year-by-year data shows volume more than 
doubling from 2015 to 2025 despite sanctions rising from 4 to 22, with 
dips in 2020 and 2023. The 2020 dip occurred while sanctions slightly 
fell, so sanctions do not explain it; a possible external factor such as

countries are never sanctioned, which affects group comparisons.

```python
# 3) Which countries prefer which supplier more
```

```python
supplier_analysis=semiconductor_trade_flows.groupby(['importer_country','primary_supplier'])['volume_units'].sum().reset_index()
```

```python
supplier_analysis['country_total']=supplier_analysis.groupby(['importer_country'])['volume_units'].transform(sum)
```

```python
supplier_analysis['percentage_share']=supplier_analysis['volume_units']/supplier_analysis['country_total']*100
```

```python
supplier_analysis['top_unit_volume']=supplier_analysis.groupby(['importer_country'])['volume_units'].transform(max)
```

```python
max_country=supplier_analysis.groupby(['importer_country'])['volume_units'].idxmax()
top_supplier=supplier_analysis.loc[max_country]
top_supplier.sort_values('importer_country')
```

```python
final_result=top_supplier[['importer_country','primary_supplier','percentage_share']]
final_result.sort_values(by='percentage_share',ascending=False)
```

```python
rank_suppliers=supplier_analysis.sort_values(['importer_country','volume_units'],ascending=[True,False])
```

```python
rank_suppliers['rank']=(rank_suppliers.groupby('importer_country').cumcount()+1)
```

```python
second_rank=rank_suppliers[rank_suppliers['rank']==2][['importer_country', 'primary_supplier', 'percentage_share']]
```

```python
second_rank=second_rank.rename(columns={'primary_supplier':'second_top_suplier','percentage_share':'second_top_percentage'})
```

```python
final_result=final_result.merge(second_rank,on='importer_country',how='left')
```

```python
final_result['Difference_of_share']=final_result['percentage_share']-final_result['second_top_percentage']
```

```python
final_result.sort_values(by='Difference_of_share',ascending=True)
```

```python
plt.figure(figsize=(10,5))
plt.barh(final_result['importer_country'],final_result['percentage_share'])
plt.xlabel('Percentage of Share of Top supplier') 
plt.ylabel('Impoting Country')
plt.title('Country and their top supplier percentage')
plt.gca().invert_yaxis()
plt.show()
```

### Interpretation:
No importing country shows a strong preference for one supplier. The 
top supplier's share is only 7.6% to 16.6% everywhere, and in six of 
eleven countries the gap between first and second place is under 1 
percentage point, which I treat as effectively tied. Taiwan is the 
exception, with AMD (16.0%) well ahead of Intel Arc (8.6%). 
Limitations: volume received does not show whether a supplier was 
chosen, supplier labels may split one company across product lines, 
and the data is partly synthetic.
